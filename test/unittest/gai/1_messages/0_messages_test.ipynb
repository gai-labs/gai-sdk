## Message Body Type Registration

### a) Loading a Built-in Message Body Type

```python
from gai.messages.typing import DefaultBodyPydantic,get_message_cls
MessagePydantic = get_message_cls()

message = MessagePydantic(
    body = DefaultBodyPydantic(
        content="hello"
    )
)
print(message)
```

### b) Loading a Custom Message Body Type

```python
from gai.messages.typing import get_message_cls,register_body
from pydantic import BaseModel
from typing import Literal

@register_body
class PingBodyPydantic(BaseModel):
    type: Literal["Ping"] = "Ping"
    content: str

MessagePydantic = get_message_cls()
message = MessagePydantic(
    body = PingBodyPydantic(content="Ping")
)
print(message)
```

## Message Serialization

### a) Serializing a Built-in Message Body Type

```python
from gai.messages.typing import MessagePydantic,StateBodyPydantic
state_json = {
    'id': 'a0e5f98c-f6eb-47de-a6e2-387510d970f9', 
    'header': {
        'sender': 'User', 
        'recipient': 'Assistant', 
        'timestamp': 1752213128.0972457, 
        'order': 0
    }, 'body': {
        'type': 'state', 
        'state_name': 'CHAT', 
        'step_no': 1, 
        'content_type': 'text', 
        'role': 'user', 
        'content': 'Tell me a one paragraph story'
    }
}
message = MessagePydantic(**state_json)
print(message)
assert isinstance(message.body,StateBodyPydantic)
```

### b) Serializing a Custom Message Body Type

```python
from typing import Literal
from pydantic import BaseModel
from gai.messages.typing import get_message_cls,register_body

@register_body
class PingBodyPydantic(BaseModel):
    type: Literal["Ping"] = "Ping"
    content: str

MessagePydantic = get_message_cls()

jsoned = {
    'id': 'a0e5f98c-f6eb-47de-a6e2-387510d970f9',
    'header': {
        'sender': 'User',
        'recipient': 'Assistant',
        'timestamp': 1752213128.0972457,
        'order': 0
    },
    'body': {
        'type': 'Ping',
        'content': 'Ping'
    }
 }
message = MessagePydantic(**jsoned)
print(message)
assert isinstance(message.body,PingBodyPydantic)
```
